# Display Outputs

Import required modules.

```python
import sys
from pathlib import Path
project_root = Path().resolve().parent
sys.path.append(str(project_root))

import random
from src.processing import read_write_data
```

Define some functions to view the patients journey and notes.

```python
def get_journey(clinical_notes_df, return_ids = True, display_journey = True):

    if return_ids:
        return [note_id for note_id in clinical_notes_df["clinical_note_id"]]

def display_note(clinical_notes_df, note_id):

    note_row = clinical_notes_df[clinical_notes_df["clinical_note_id"] == note_id]
    note = note_row["clean_note_text"].iloc[0]
    
    print(note)

def print_admission_header(admission_id):
    print("\n" + "=" * 100)
    print(f"ADMISSION ID: {admission_id}")
    print("=" * 100 + "\n")


def print_note_header(i, note_id):
    print("\n" + "-" * 80)
    print(f"NOTE {i} | NOTE ID: {note_id}")
    print("-" * 80 + "\n")
```

This notebook allows quick and easy viewing of patient journeys and clincial notes.

```python
clinical_notes = read_write_data("clinical_notes", "read")

# Extract out ID dataframe from clinical notes.
id_df = clinical_notes[["person_id", "admission_id", "encounter_id", "clinical_note_id"]].drop_duplicates()
```

```python
# Extract out admissions ids
admission_ids = id_df.admission_id.unique().tolist()
```

Fetch clinical notes from a certain admission and display their notes.

```python
N = 1
selected_admission_ids = random.sample(
    admission_ids,
    k=min(N, len(admission_ids))
)

for admission_id in selected_admission_ids:

    print_admission_header(admission_id)

    clinical_notes_v = clinical_notes[
        clinical_notes["admission_id"] == admission_id
    ]

    note_ids = get_journey(clinical_notes_v)

    for i, note_id in enumerate(note_ids, start=1):
        print_note_header(i, note_id)
        display_note(clinical_notes_v, note_id)
```
